## Architecture Comparison

| Architecture | Complexity | Use Case | Key Feature |
|--------------|------------|----------|-------------|
| **Simple RAG** | 1/5 | General purpose | Query → Retrieve → Generate |
| **Memory RAG** | 2/5 | Chatbots, Support | Conversation history |
| **Branched RAG** | 3/5 | Multi-domain | Parallel sub-queries |
| **HyDe** | 3/5 | Ambiguous queries | Hypothetical docs |
| **Adaptive RAG** | 4/5 | Mixed workloads | Intelligent routing |
| **CRAG** | 4/5 | High-accuracy | Relevance + web fallback |
| **Self-RAG** | 5/5 | Quality-critical | Self-critique loop |
| **Agentic RAG** | 5/5 | Complex reasoning | Autonomous agents |
| **Contextual RAG** | 3/5 | Technical docs | Context-augmented chunks |
| **Fusion RAG** | 3/5 | Best ranking | Reciprocal Rank Fusion |
| **SQL RAG** | 4/5 | Analytics/BI | Natural Language to SQL |
| **GraphRAG** | 5/5 | Knowledge graphs | Entity relationships + multi-hop |
| **Multimodal RAG** | 4/5 | Images + text | Vision model + OCR |
| **Hybrid + Rerank** | 3/5 | Identifiers, jargon | BM25 + dense (RRF) + cross-encoder |
| **Parent / Multi-Vector** | 3/5 | Chunk-size trade-offs | Search small or generated vectors, return full context |

**Legend:**
- 1/5 = Simple (10-15 min)
- 3/5 = Moderate (15-20 min)
- 5/5 = Advanced (25-35 min)
- = New in v1.1.0

## Need Help?

### Common Issues

1. **"Module 'shared' not found"**
   - Ensure you're running from the correct directory
   - Run: `sys.path.append('..')` at the start of notebook

2. **"OpenAI API key not found"**
   - Create `.env` file in project root
   - Add: `OPENAI_API_KEY=sk-proj-...`
   - Restart Jupyter kernel

3. **"Vector store not found"**
   - Run `fundamentals/02_embeddings_comparison.ipynb` first
   - This creates the vector stores in `data/vector_stores/`

4. **HuggingFace model download slow**
   - First run downloads ~90MB
   - Subsequent runs use cached model
